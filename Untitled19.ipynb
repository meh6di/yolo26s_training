```python
import zipfile

# ✏️ Update filename if yours differs
with zipfile.ZipFile('Waste_Detection.v1-new.yolo26.zip', 'r') as z:
    z.extractall('/content/dataset')
```

```python
import yaml

with open('/content/dataset/data.yaml') as f:
    cfg = yaml.safe_load(f)

cfg['path']  = '/content/dataset'
cfg['train'] = 'train/images'
cfg['val']   = 'valid/images'
cfg['test']  = 'test/images'

with open('/content/data.yaml', 'w') as f:
    yaml.dump(cfg, f)
```

```python
from ultralytics import YOLO

model = YOLO('yolo26s.pt')

results = model.train(
    data='/content/data.yaml',
    epochs=300,
    imgsz=640,
    batch=8,
    patience=0,
    device=0,
    augment=True,
    mosaic=1.0,
    fliplr=0.5,
    degrees=10.0,
    project='/content/runs',
    name='waste_detection',
)
```

```python
best_pt = str(results.save_dir) + '/weights/best.pt'
model_best = YOLO(best_pt)

metrics = model_best.val(data='/content/data.yaml', split='test', imgsz=640, device=0)
print(f'mAP50:     {metrics.box.map50:.4f}')
print(f'mAP50-95:  {metrics.box.map:.4f}')
print(f'Precision: {metrics.box.mp:.4f}')
print(f'Recall:    {metrics.box.mr:.4f}')
```
